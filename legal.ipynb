```python
frutas = ["maça", "banana", "abacaxi"]
print(frutas)
fruta_pedido = input("Adicione a sua fruta favorita!: ")
# .append so adiciona a lista
frutas.append(fruta_pedido)
print(frutas)
# .insert escolhe o lugar que deseja colocar o caractere (insert)
frutas.insert(0, fruta_pedido)
print(frutas)
# .pop retira o index do caractere
fruta_removida = frutas.pop(1)
print(f"A fruta removida foi {fruta_removida}")
# .clear limpa a lista inteira
# .remove retira o caractere pelo nome
# .sort organiza em ordem alfabetica e numerica
# caso coloque .sort(reverse=true) a ordem fica ao contrario
# .index mostra a posição do caractere como 1, 2, 3...
# .lenght conta quantos itens ha no total
frutas.sort()
print(frutas)
```

```python
cores = ("azul", "vermelho", "branco")
# neste caso o .sort nao funcionaria porque esta entre parenteses (nao é uma lista)
# para transformar em lista so usar list
cores_lista = list(cores)
print(cores_lista)
# para transformar em tuple (parenteses) so usar o tuple
cores = tuple(cores_lista)
print(cores)
```

```python
#dicionario
cachorro = {"raça":"pastor","nome":"lina", "idade":"3"}
print(cachorro["nome"])
# ou usa o .get para conseguir algo especifico como nome
print(cachorro.get("nome"))
print(cachorro.get("tamanho", "tamanho não encontrado"))
# para adicionar index no dicionario so fazer
cachorro["peso"] = "fortinha"
print(cachorro)
```

```python
numb = [1, 2, 3, 4]
print(numb)
numb.append(6)
print(numb)
numb.pop(1)
print(numb)
numb.remove(4)
numb.insert(3, 40)
print(numb)
numb.sort(reverse = True)
print(numb)
```
