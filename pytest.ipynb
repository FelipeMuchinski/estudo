```python
%%ipytest

# Exercício 1: Atribuição de nome e vida
def test_deve_armazenar_nome_e_vida_corretamente():
    p = Personagem("Felipe", 100)
    assert p.nome == "Felipe"
    assert p.vida == 100
    assert p.vida_maxima == 100

# Exercício 2 e 3: Receber dano e limite zero
def test_deve_subtrair_vida_ao_receber_dano():
    p = Personagem("Giovanna", 100)
    p.receber_dano(30)
    assert p.vida == 70

def test_vida_nao_deve_ficar_negativa():
    p = Personagem("Henrique", 20)
    p.receber_dano(50)
    assert p.vida == 0

# Exercício 4: Dano inválido
def test_deve_gerar_erro_para_dano_negativo():
    p = Personagem("Clau", 100)
    with pytest.raises(ValueError):
        p.receber_dano(-10)
    assert p.vida == 100

# Exercício 5 e 6: Curar e limite máximo
def test_deve_aumentar_vida_ao_curar():
    p = Personagem("Muchinski", 100)
    p.vida = 40
    p.curar(20)
    assert p.vida == 60

def test_cura_nao_deve_ultrapassar_vida_maxima():
    p = Personagem("Muchinski", 100)
    p.vida = 90
    p.curar(30)
    assert p.vida == 100

# Exercício 7: Cura inválida
def test_deve_gerar_erro_para_cura_negativa():
    p = Personagem("dragao", 100)
    with pytest.raises(ValueError):
        p.curar(-5)

# Exercício 8: Personagem derrotado
def test_personagem_esta_derrotado():
    p = Personagem("cachorro", 10)
    assert p.esta_derrotado() is False
    p.receber_dano(10)
    assert p.esta_derrotado() is True

# Exercício 9: Ataque entre personagens
def test_personagem_deve_atacar_outro():
    atacante = Personagem("Heroi", 100)
    alvo = Personagem("Vilao", 100)
    atacante.atacar(alvo, 40)
    assert alvo.vida == 60

# Exercício 10: Combate em turnos
def test_combate_deve_ter_um_vencedor():
    p1 = Personagem("Guerreiro", 50)
    p2 = Personagem("Mago", 20)

    # Simulação
    while not p1.esta_derrotado() and not p2.esta_derrotado():
        p1.atacar(p2, 10)
        if not p2.esta_derrotado():
            p2.atacar(p1, 10)

    assert p2.esta_derrotado() is True
    assert p1.vida > 0
```

```python
%%ipytest
class Personagem:
    def __init__(self, nome, vida):
        self.nome = nome
        self.vida_maxima = vida
        self.vida = vida

    def receber_dano(self, quantidade):
        if quantidade < 0:
            raise ValueError("O dano não pode ser negativo.")

        self.vida -= quantidade
        if self.vida < 0:
            self.vida = 0

    def curar(self, quantidade):
        if quantidade < 0:
            raise ValueError("A cura não pode ser negativa.")

        self.vida += quantidade
        if self.vida > self.vida_maxima:
            self.vida = self.vida_maxima

    def esta_derrotado(self):
        return self.vida == 0

    def atacar(self, alvo, dano):
        alvo.receber_dano(dano)
```
