```python
def num(numero):
  if numero % 2 == 0:
    return True
  else:
    return False
# o que eu quero fazer com a função é verificar se o numero é par ou impar, ou seja posso usar ela ali com mais facilidade
if num(2):
  print("é par")
else:
  print("é impar")
```

```python
#o * faz o usuario usar quantos argumentos ele quiser na função
def somar_lista(*numeros):
  print(numeros)
  resultado = 0
  for numero in numeros:
    resultado += numero
  return resultado

soma_dalista = somar_lista(1, 2, 5, 745, 20)
print(f"o resultado é {soma_dalista}")



```

```python
def calculo(*numeros):
  qtd = len(numeros)
  soma = 0
  for numero in numeros:
    soma += numero
  media = soma / qtd
  return media
resultado = calculo(3, 5, 7, 78)
print(f"a media é {resultado}")
```

```python
# dois asteriscos (**) fazem uma tupla fora da função virar dicionario
def documentos(**info):
  for c, v in info.items():
    print(f"{c}:{v}")

documentos(nome="felipe", idade="19", lina="bonita" )
```

```python
def contagem(numeros):
  for num in range(numeros, -1, -1):
    print(f"{num}")

contagem(5)
#como fazer contagem regressiva
```

```python
def maior_num(numeros_na_lista):
  maior_num = max(numeros_na_lista)
  return maior_num

lista = [1, 4, 6, 78, 8]
maiornum_dalista = maior_num(lista)
print(f"o maior numero da lista é {maiornum_dalista}")

```
