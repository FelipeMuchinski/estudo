```python
numeros = [1, 3, 5, 7, 8, 32, 6, 0, 9]
for numero in numeros:
  if numero % 2 != 0:
    continue
  print(f"o numero {numero} é par")


dados = {"nome":"felipe", "idade":19, "beleza":"bonito"}
#usei c para chave e v para valor
for c, v in dados.items():
  # .items retorna mais de um valor do dicionario
  print(f"{c}: {v}")
#para pegar so os valores podemos usar o .values
for v in dados.values():
  print(f"{v}")
# ja para as chaves usamos o .keys
for c in dados.keys():
  print(f"{c}")
```
