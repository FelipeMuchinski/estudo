```python
import random

class Lutador:
  def __init__(self, nome, classe, forca, agilidade, vida, defesa) -> None:
    if vida < 0:
      raise ValueError("A vida não pode ser negativa!")
    self.nome = nome
    self.classe = classe
    self.forca = forca
    self.agilidade = agilidade
    self.vida = vida
    self.defesa = defesa
    self.guarda = False

  def atacar(self, alvo):
    if alvo is None:
      print(f"{self.nome} testou seu ataque")
      return
    dano = self.forca - (alvo.defesa if not alvo.guarda else alvo.defesa * 1.5)
    dano_real = max(0, int(dano))

    print(f" ---> {self.nome} ataca {alvo.nome}!!!")


    if alvo.esquivar():
      print(f"{alvo.nome} desviou!!")
    else:
      alvo.pontos_vida(-dano_real)
      print(f"{alvo.nome} recebeu {dano_real} de dano!")
      alvo.guarda = False

  def defender(self):
    print(f"{self.nome} está em guarda!")
    self.guarda = True

  def esquivar(self):
    chance = random.randint(0, 100)
    return chance < self.agilidade

  def pontos_vida(self, valor):
    self.vida += valor
    if self.vida < 0:
      self.vida = 0
    print(f"A vida de {self.nome}: {self.vida}")

  def __str__(self):
    return f"Lutador: {self.nome} | Saúde: {self.vida} | Forca: {self.forca} | Defesa: {self.defesa}"
```

```python
lutador1 = Lutador("muchinski", "barbaro", 90, 50, 80, 80)
lutador2 = Lutador("zaidan", "mago", 50, 70, 60, 40)
```

```python
print("Início da Simulação Manual:")
print(lutador1)
print(lutador2)
print("-" * 30)


lutador1.atacar(lutador2)
lutador2.defender()
lutador1.atacar(lutador2)
lutador2.atacar(lutador1)


print("-" * 30)
print("Estado Final:")
print(lutador1)
print(lutador2)
```

```python
import random

def simular_combate(l1, l2, max_rodadas=20):
    print(f"\n{"="*10} O COMBATE COMEÇOU: {l1.nome} vs {l2.nome} {"="*10}")

    rodada = 1

    while l1.vida > 0 and l2.vida > 0 and rodada <= max_rodadas:
        print(f"\n--- RODADA {rodada} ---")

        for atacante, defensor in [(l1, l2), (l2, l1)]:
            if atacante.vida > 0:

                acao = random.choice(["atacar", "defender"])

                if acao == "atacar":
                    atacante.atacar(defensor)
                else:
                    atacante.defender()

            if defensor.vida <= 0:
                break

        rodada += 1

    print(f"\n{"="*15} FIM DE COMBATE {"="*15}")
    print(f"Rodadas totais: {rodada - 1}")

    if l1.vida > l2.vida:
        print(f"VENCEDOR: {l1.nome.upper()}!")
    elif l2.vida > l1.vida:
        print(f"VENCEDOR: {l2.nome.upper()}!")
    else:
        print("EMPATE! Ambos continuam de pé (ou caíram juntos).")

    print(f"\nEstado final dos lutadores:")
    print(l1)
    print(l2)

muchinski = Lutador("Muchinski", "Barbaro", 90, 50, 80, 80)
zaidan = Lutador("Zaidan", "Mago", 50, 70, 60, 40)

simular_combate(muchinski, zaidan)
```
