```python
preco_pao = 0.99

# Corrigido a ordem do float(input)
saldo = float(input("Digite seu saldo: "))

if saldo < preco_pao:
    print("Desculpe, você não tem dinheiro nem para um pão.")
else:
    # Convertendo a quantidade para número inteiro
    quantidade_pao = int(input("Digite quantos pães você deseja: "))
    resultado = quantidade_pao * preco_pao

    if resultado > saldo:
        print("Desculpe, saldo insuficiente para essa quantidade.")
    else:
        # Usando f-string para exibir os valores corretamente
        print(f"Perfeito! Deu {resultado:.2f} reais")

        sobra = saldo - resultado
        print(f"Aqui está seu troco de {sobra:.2f} reais")
```
