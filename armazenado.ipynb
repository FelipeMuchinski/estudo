```python
#preço do pao como variavel e um cliente deseja comprar, caso apos ele digitar o saldo confere se tem dinheiro o suficiente e calcula

preco_pao = float(0.99)
saldo = float(input("Digite seu saldo:"))
if saldo < preco_pao:
  print("Desculpe não aceitamos fiado")
else:
      quantidade_pao = int(input("Digite quantos pães voce deseja:"))
resultado = quantidade_pao * preco_pao
if resultado > saldo:
  print("Desculpe, saldo insuficiente")
else:
  print(f"Perfeito! Deu {resultado:.2f} reais")
  sobra = saldo - resultado
  print(f"Aqui esta seu troco de {sobra:.2f} reais")




```
