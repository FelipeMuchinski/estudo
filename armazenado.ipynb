```python
def normalizar_nome(nome):
    """Remove espaços extras no início, fim e entre as palavras."""
    # O split() sem argumentos remove qualquer sequência de espaços em branco
    # O join() junta as palavras com apenas um espaço entre elas
    return " ".join(nome.split())

def remover_preposicoes(nome):
    preposicoes = ["de", "da", "do", "dos", "das", "e"]
    palavras = nome.split()
    # Mantém a palavra se a versão minúscula dela não estiver na lista de preposições
    filtrado = [a for a in palavras if a.lower() not in preposicoes]
    return " ".join(filtrado)

def extrair_primeiro_nome(nome):
    palavras = nome.split()
    return palavras[0].lower()

def extrair_sobrenome(nome):
    palavras = nome.split()
    return palavras[-1].lower()

def formatar_curso(curso):
    return curso[:3].lower()

def gerar_login(nome, curso, ano):
    nome_limpo = normalizar_nome(nome)
    nome_sem_prep = remover_preposicoes(nome_limpo)
    primeiro = extrair_primeiro_nome(nome_sem_prep)
    sobrenome = extrair_sobrenome(nome_sem_prep)
    curso_sigla = formatar_curso(curso)
    return f"{primeiro}.{sobrenome}.{curso_sigla}.{ano}"

nome_aluno = input("Nome: ")
ano_ingresso = str(input("Ano: "))
curso_nome = input("Curso: ")

login1 = gerar_login(nome_aluno, curso_nome, ano_ingresso)
print(f"Login 1: {login1}")

```
